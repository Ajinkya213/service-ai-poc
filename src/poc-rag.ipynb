```python
import os
from byaldi import RAGMultiModalModel 
import warnings 
warnings.filterwarnings('ignore')
```

```python
#Define the model
model='vidore/colpali-v1.3'
RAGmodel=RAGMultiModalModel.from_pretrained(model,verbose=0)
```

```python
RAGmodel.index(
    input_path='/input/prusa-mks3-5-pdf-data',
    index_name='/poc_vector_data',
    store_collection_with_index=True,
    overwrite=True,
)
```

```python
# Define the query to search
query='How to install the steel ball?'
results=RAGmodel.search(query, k=2)
```

```python
import base64
image_byte=[]
for i in range(0,2):
    image_byte.append(base64.b64decode(results[i].base64))
len(image_byte)
```

```python
file_names=[]
for i,image in enumerate(image_byte):
    filename = f'result_image{i+1}.jpg'
    file_names.append(filename)
    with open(filename, 'wb') as f:
      f.write(image)
print(file_names)
```

```python
from IPython.display import Image
for i in file_names:
    display(Image(i))
```

```python
from PIL import Image
images=[Image.open(path) for path in file_names]

prompt= f"""You are an AI model specialized in image analysis and question answering. Below are one or more images and a query based on them.
Carefully analyze the image(s), extract relevant information, and provide a well-structured response to the query.

Query:
{query}

Instructions:
1. Describe the key elements or patterns observed in the image(s).
2. Use deductive reasoning to interpret the visuals in relation to the query.
3. Provide a clear and concise answer to the query with justification.
4. If applicable, highlight any assumptions or limitations.

Format your response as follows:
- Answer to Query: [Final answer]
- Reasoning: [Explain how you arrived at the answer]
- Document: [Include the document name as well as page number of the document]
"""
```

```python
from dotenv import load_env
load_env()
gemini_key=os.environ.get('GEMINI_API')
```

```python
import google.generativeai as genai 
from IPython.display import Markdown 
genai.configure(api_key=gemini_key)
gen_model=genai.GenerativeModel('gemini-1.5-flash')

chat=gen_model.start_chat()
response=chat.send_message([prompt,*images])

answer=response.text
display(Markdown(answer))
```
